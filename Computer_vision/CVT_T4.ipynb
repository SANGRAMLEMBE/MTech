```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# ===============================
# Upload Image
# ===============================

filename = ('/content/photo-1611250282006-4484dd3fba6b.jpeg')

img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ===============================
# Parameters
# ===============================
levels = 5   # Number of pyramid levels

# ===============================
# Build Gaussian Pyramid
# ===============================
pyramid = [img]

for i in range(1, levels):
    blurred = cv2.GaussianBlur(pyramid[i-1], (5,5), 1)
    downsampled = blurred[::2, ::2]   # Downsample by factor 2
    pyramid.append(downsampled)

# ===============================
# Display Pyramid
# ===============================
plt.figure(figsize=(15,8))

for i in range(levels):
    plt.subplot(2, (levels+1)//2, i+1)
    plt.imshow(pyramid[i])
    plt.title(f"Level {i}\n{pyramid[i].shape[1]}x{pyramid[i].shape[0]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
