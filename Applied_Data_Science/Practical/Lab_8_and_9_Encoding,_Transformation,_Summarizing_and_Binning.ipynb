```python
import pandas as pd
import numpy as np
```

```python
#Reading the file

df=pd.read_csv('/content/Superstore.csv', encoding='latin1')
display(df.head())
```

```python
df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
if 'Ship Date' in df.columns:
    df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')
```

```python
df.info()
```

```python
# creating profit margin column in df
df['ProfitMargin'] = np.where(df['Sales'] == 0, np.nan, df['Profit'] / df['Sales'])
df.head()
```

```python
# Average ProfitMargin per Region
avg_pm_region = df.groupby('Region')['ProfitMargin'].mean().sort_values(ascending=False)
print("\nAverage ProfitMargin by Region:")
print(avg_pm_region)
```

```python
df['OrderMonth'] = df['Order Date'].dt.to_period('M').dt.to_timestamp()
```

```python
df.info()
```

```python
df.head()
```

```python
# Compute total Sales per Region x Month
monthly_sales = (
    df
    .groupby(['Region', 'OrderMonth'], as_index=False)['Sales']
    .sum()
    .sort_values(['Region', 'OrderMonth'])
)
print("\nTotal Sales per Region x Month:")
print(monthly_sales)
```

```python
# For each region, compute previous month's total sales (shift)
monthly_sales['Sales_prev_month'] = monthly_sales.groupby('Region')['Sales'].shift(1)
monthly_sales.head()
```

```python
# Merge Sales_prev_month back into original df (join on Region + OrderMonth)
df = df.merge(
    monthly_sales[['Region', 'OrderMonth', 'Sales_prev_month']],
    on=['Region', 'OrderMonth'],
    how='left'
)

df.head()
```

```python
df.info()
```

```python
# Show a few rows
print("\nExample rows with ProfitMargin and Sales_prev_month:")
print(df[['Order ID','Order Date','Region','Sales','Profit','ProfitMargin','OrderMonth','Sales_prev_month']].head(8))
```

```python
# --- 2. Encoding ---
```

```python
# One-hot encode Segment
if 'Segment' in df.columns:
    seg_dummies = pd.get_dummies(df['Segment'], prefix='Segment')
    df = pd.concat([df, seg_dummies], axis=1)
    print("\nOne-hot columns added:", list(seg_dummies.columns))
```

```python
# Target encode Region by mean ProfitMargin (map mean ProfitMargin per region into new column)
region_pm_mean = df.groupby('Region')['ProfitMargin'].mean()
df['Region_TargetEnc'] = df['Region'].map(region_pm_mean)
```

```python
print("\nRegion -> mean ProfitMargin mapping:")
print(region_pm_mean)
```

```python
# --- 3. Standardization / Scaling ---
```

```python
numeric_cols = ['Sales', 'Quantity', 'Discount', 'ProfitMargin']
# Confirm columns exist; create scaled columns for only existing ones
existing_numeric = [c for c in numeric_cols if c in df.columns]
existing_numeric
```

```python
# Compute z-score with ddof=0 (population std); if you prefer sample std use ddof=1
for col in existing_numeric:
    mean = df[col].mean(skipna=True)
    std = df[col].std(ddof=0, skipna=True)
    if std == 0 or np.isnan(std):
        # Avoid divide by zero: produce zeros if all equal
        df[f'{col}_z'] = 0.0
    else:
        df[f'{col}_z'] = (df[col] - mean) / std
```

```python
# Count "high-sales outlier" orders where Sales_z > +2
if 'Sales_z' in df.columns:
    high_sales_outliers = df[df['Sales_z'] > 2]
    n_high_sales_outliers = len(high_sales_outliers)
else:
    n_high_sales_outliers = 0

print(f"\nNumber of high-sales outlier orders (Sales_z > +2): {n_high_sales_outliers}")
```

```python
# Optionally: display top 10 high-sales orders
if n_high_sales_outliers > 0:
    print("\nTop high-sales outlier orders (by Sales_z):")
    display_cols = ['OrderID' if 'OrderID' in df.columns else 'Order Id','OrderDate','Region','Sales','Sales_z','Profit','ProfitMargin']
    display_cols = [c for c in display_cols if c in df.columns]
    print(high_sales_outliers.sort_values('Sales_z', ascending=False)[display_cols].head(10))
```

```python
# --- 4. Binning (discretization)
```

```python
# Create SalesBin into 4 quartiles labeled Low, Medium, High, Very High
# Use qcut; drop NA Sales first if any
df_nonnull_sales = df[df['Sales'].notna()].copy()
labels = ['Low', 'Medium', 'High', 'Very High']
try:
    df_nonnull_sales['SalesBin'] = pd.qcut(df_nonnull_sales['Sales'], q=4, labels=labels)
except ValueError as e:
    # if qcut fails (e.g., too many identical values), fallback to cut with quantiles computed manually
    quantiles = df_nonnull_sales['Sales'].quantile([0, .25, .5, .75, 1.0]).values
    df_nonnull_sales['SalesBin'] = pd.cut(df_nonnull_sales['Sales'], bins=quantiles, labels=labels, include_lowest=True)
```

```python
# Compute average ProfitMargin for each Category × SalesBin
if 'Category' in df_nonnull_sales.columns:
    pivot_pm = pd.pivot_table(
        df_nonnull_sales,
        index='Category',
        columns='SalesBin',
        values='ProfitMargin',
        aggfunc='mean'
    )
    print("\nAverage ProfitMargin by Category × SalesBin (rows=Category, columns=SalesBin):")
    print(pivot_pm)
else:
    pivot_pm = None
    print("\nNo 'Category' column found in data — cannot compute pivot.")
```

```python
# Interpretation helper: show top categories within 'Very High' bin sorted by ProfitMargin
if pivot_pm is not None and 'Very High' in pivot_pm.columns:
    top_in_very_high = pivot_pm['Very High'].sort_values(ascending=False)
    print("\nCategories sorted by ProfitMargin within 'Very High' SalesBin:")
    print(top_in_very_high)
```
