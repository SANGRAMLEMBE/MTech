1) Total sales per country
*  Create a Sales column = Quantity * Price. Compute total sales per Country.

```python
df["Sales"]=df["Quantity"]*df["Price"]
df["Sales"][:3]
```

```python
total_sales_per_country = df.groupby("Country")["Sales"].sum()
total_sales_per_country
```

2) Groupby multiple aggregations
* For each Country, compute:
* total sales
* number of unique customers
* total quantity sold

```python
total_sales_per_country = df.groupby("Country")["Sales"].sum()
total_sales_per_country
```

3)Top 10 products by revenue
* Group by StockCode and find top 10 items by total sales.

```python
df.groupby("StockCode")["Sales"].sum().sort_values(ascending=False).head(10)
```

4)Monthly sales trend
* Group by month (using InvoiceDate) and compute monthly total sales.

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df["month"]=df["InvoiceDate"].dt.month
```

```python
monthly_sales = df.groupby('month')['Sales'].sum()
display(monthly_sales)
```

5)Average order value by customer
* For each CustomerID, compute total revenue and number of invoices; then compute average
order value (AOV = revenue / #invoices).

```python

pivot_table = df.pivot_table(index='Customer ID', values=['Sales', 'Invoice'], aggfunc={'Sales': 'sum', 'Invoice': 'count'})

pivot_table["AOV"] = pivot_table["Sales"] / pivot_table["Invoice"]
pivot_table
```

6)Pivot table: Country × Month average sales
* Pivot showing mean monthly sales per Country (index: Country, columns: month-year).

```python
pivot_cunt_month = df.pivot_table(index='Country', columns='month', values='Sales', aggfunc='mean').fillna(0)

pivot_cunt_month
```

7)Pivot with multiple aggfuncs
* Pivot with index Country, columns StockCode (top N only), values Sales, aggfunc [np.sum,
'count'] (or limit columns to top items).

```python
pivot_cunt_stc = df.pivot_table(index='Country', columns='StockCode', values='Sales', aggfunc=['sum', 'count']).fillna(0)
pivot_cunt_stc = pivot_cunt_stc.sort_values(by=('count', 'POST'), ascending=False)
display(pivot_cunt_stc)
```

8)Groupby with filter

* Among orders with Quantity < 0 (returns/cancellations), compute the number of return
transactions per Country.

```python
df.groupby("Country")["Quantity"].apply(lambda x: (x < 0).sum())
```

9)Crosstab: Country vs High-value order
* Define a high-value order as Sales > 500. Use pd.crosstab to show counts of high-value vs
low-value orders per Country, and include margins.

```python
df['High_Value_Order'] = df['Sales'] > 500

crosstab_country_high_value = pd.crosstab(df['Country'], df['High_Value_Order'], margins=True)
display(crosstab_country_high_value)
```

10)Customer repeat-rate using crosstab
* Create a column is_repeat indicating whether the customer has more than one invoice in the
dataset. Use pd.crosstab to show counts of repeat vs one-time customers by Country.

```python
invoice_counts = df.groupby('Customer ID')['Invoice'].nunique()
df['is_repeat'] = df['Customer ID'].map(invoice_counts) > 1
crosstab_country_repeat = pd.crosstab(df['Country'], df['is_repeat'], margins=True)
display(crosstab_country_repeat)
```
